## **Ethics of Web Scraping**

### 1. Is it Legal and Allowed?
* **Public Access:** You can generally scrape public data if you don't crash the server (In most cases).
* **Private Rights:** <mark>Websites are private property and can block you at any time.</mark> If the appearance of your scraper is like a bot, it is very easy to block you. Remember the websites and platforms are thier own private entities and can do anything. DO NOT scrape LinkedIN 😀
* **Robots.txt:** <mark>Always check this file first</mark> to see the site's specific scraping rules. Let me show you show to check!

### 2. Common Obstacles
* **400+ Errors:** These are warnings to slow down before your IP is banned. Let us look at all warnings. Understanding some **#400** status codes will be helpful! 200 is a green light.
* **IP Bans:** Professional scrapers use "rotating proxies" to hide, but these can be expensive.
* **Bot Detection:** Sites look for "superhuman" speed (e.g., clicking 600 times a minute) and trigger CAPTCHAs.(We must always be respectful of the cite and not crash the server. In many cases, in the robots.txt file, they also mention the crawl delay. In any case, always be polite, if you have and not in a hurry, add wait of few seconds everynow to mimick a human). It is also possible to bypass captchas but, we will cover in this class. If your topic requires this/you are interested, we can look at in the exercises.
* **Mimicking Humans:** Using automated browsers (like Selenium) helps bypass detection by masking your browser as a human instead of a bot. We can also use our own cookies to trick the browser into thinking that it is us!

### 3. The Better Way: APIs
* **The "Front Door":** <mark>Many major sites provide APIs to give you data directly</mark>. Of course now we in the time of **"PostAPIcalypse"** (*Bruns, A. (2019). After the ‘APIcalypse’ Social Media Platforms and Their Fight against Critical Scholarly Research*)
* **Why use them:** They are easier to use, well-documented, and authorized by the website. If you have access from the "front door", i.e., the site gives you the keys, never break in through the window 😀

### 4. Imp Rules
* **Be Kind:** Don't overload server resources, keep the concepts of data minimalism in mind, if you can scrape it but, do not need to, do not scrape it.
* **Be Ethical:** Avoid scraping personal data (like emails) for spamming.
* **Don't force it:** If an API exists, use it instead of building a custom scraper.

## Let us look at some robots.txt

To check the rules for any website, simply add `/robots.txt` to the end of the root URL (e.g., `https://www.google.com/robots.txt`).

### What to look for:
* **User-agent:** Specifies which bot the rule applies to (e.g., `*` means all bots).
* **Disallow:** <mark>Lists the directories or pages that you are NOT permitted to scrape or crawl.</mark>
* **Allow:** Explicitly mentions parts of the site that are open to bots.
* **Crawl-delay:** A value in seconds indicating how long a bot should wait between successive requests to avoid overloading the server.

````python
from re import findall
import requests
from bs4 import BeautifulSoup

url = "https://quotes.toscrape.com/page/1/"
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

quotes = soup.find_all("div", class_="quote")

for i in quotes:
  text = i.find("span", class_="text").get_text()
  author = i.find("small", class_="author").get_text()
  tags = i.find("a", class_="tag").get_text()
  print(text, ":" ,author, "-", tags)
````

````python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://quotes.toscrape.com/page/1/"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

quotes = soup.find_all("div", class_="quote")

data = []
for i in quotes:
    text = i.find("span", class_="text").get_text()
    author = i.find("small", class_="author").get_text()
    tag = i.find("a", class_="tag").get_text()
    data.append({"Quote": text, "Author": author, "Tag": tag})

# We have now created a dictionary using the append command.
# We can print in too
print(data)

# We can also save a csv
df = pd.DataFrame(data) # Making a pandas df
print(df)

#Saving it
df.to_csv("quotes.csv", index=False)
````

````python
import requests
from bs4 import BeautifulSoup
import os
from urllib.parse import urljoin

url = "https://ourworldindata.org/artificial-intelligence"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

# Since colab is a ve, we need to make the folder using code too!
if not os.path.exists("downloaded_images"):
    os.makedirs("downloaded_images")

images = soup.find_all("img")

# to save images as img1, img2 etc.
count = 0

for i in images:
  if "https:" in i.get("src"):
    link = i.get("src")
    #print(link)
    img_data = requests.get(link).content
    #print(img_data) # Dont print it xD

    filename = f"image_{count}.png"
    file_path = os.path.join("downloaded_images", filename)


    with open(file_path, 'wb') as f:
        f.write(img_data)
    count += 1
````

### 3. Understanding APIs and JSON

#### Why use an API?
* **The Front Door:** An API (Application Programming Interface) is a formal way for a website to share data with you.
* **Clean Data:** Instead of messy HTML, you get structured data (JSON), which is much easier to process.
* **Stability:** Unlike web scraping, which breaks if a website changes its layout, APIs are built to remain consistent.

#### What is JSON?
* **Format:** JSON (JavaScript Object Notation) is the standard language of APIs.
* **Structure:** It organizes data into key-value pairs, making it look very similar to a Python dictionary.

````python
from googleapiclient.discovery import build

API_KEY = "KEY_HERE"
youtube = build('youtube', 'v3', developerKey=API_KEY)
````

````python
def get_top_search_results(query, max_results=5):
    request = youtube.search().list(
        q=query,
        part="snippet",
        type="video",
        maxResults=max_results
    )
    response = request.execute()

    results = []
    for item in response.get('items', []):
        results.append({
            "title": item['snippet']['title'],
            "video_id": item['id']['videoId'],
            "channel": item['snippet']['channelTitle']
        })
    return results
````

````python
search_term = "Policy Analysis and Language Models"
top_videos = get_top_search_results(search_term, max_results=10)

for i, vid in enumerate(top_videos, 1):
    print(f"{i}. {vid['title']} (ID: {vid['video_id']})")
````
